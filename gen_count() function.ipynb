## write a function to calculate the count of genders from a list.

```python
def gen_count():
    
    l = eval(input("Enter a list of genders : "))

    male = []
    female = []

    for i in l:
        if i == 'M':
            male.append(i)

        elif i == 'F':
            female.append(i)

    print("Count of Males in list : ",len(male))
    print("Count of Females in list : ",len(female))

    len(l)
    
gen_count()
```
